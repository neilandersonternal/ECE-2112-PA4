```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("board2.csv")
df
```

```python
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
```

```python
df.describe()
```

```python
# **A. VISAYAS COMMUNICATION DATAFRAME**
```

```python
VisComm = df[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication')]
VisComm
```

```python
VisComm = VisComm[['Name', 'Gender', 'Math', 'Electronics', 'Average']]
VisComm
```

```python
print("Visayas Communication Dataframe")
display(VisComm)
print(f"Number of rows: {len(VisComm)}\n")
```

```python
# **B. VISAYAS FEMALE DATAFRAME**
```

```python
VisFemale = df[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female')]
VisFemale
```

```python
VisFemale = VisFemale[['Name', 'Track', 'GEAS', 'Electronics', 'Average']]
VisFemale
```

```python
print("Visayas Female DataFrame:")
display(VisFemale)
print("\nVisayas Female Students with Average >= 60:")
display(VisFemale[VisFemale['Average'] >= 60])
```

```python
# **C. CATEGORY-AVERAGE VISUALIZATION**
```

```python
track_mean = df.groupby('Track')['Average'].mean().reset_index()
 
```

```python
gender_mean = df.groupby('Gender')['Average'].mean().reset_index()
```

```python
hometown_mean = df.groupby('Hometown')['Average'].mean().reset_index()
```

```python
display(track_mean)
display(gender_mean)
display(hometown_mean)
```

```python
track_max = track_mean.max()

track_max
```

```python
gender_max = gender_mean.max()

gender_max
```

```python
hometown_max = hometown_mean.max()

hometown_max
```

```python

plt.figure(figsize=(20, 5))

plt.subplot(1, 3, 1)
plt.bar(track_mean['Track'], track_mean['Average'])
plt.title('Mean Average by Track')
plt.xlabel('Track')
plt.ylabel('Mean Average Score')

plt.subplot(1, 3, 2)
plt.bar(gender_mean['Gender'], gender_mean['Average'])
plt.title('Mean Average by Gender')
plt.xlabel('Gender')
plt.ylabel('Average')

plt.subplot(1, 3, 3)
plt.bar(hometown_mean['Hometown'], hometown_mean['Average'])
plt.title('Mean Average by Hometown')
plt.xlabel('Hometown')
plt.ylabel('Average')


    
plt.tight_layout()
plt.show()


print("The track that has the highest mean average across all tracks is:\n")
print(track_max) 
print(" ")

print("Where as for the Gender category, the gender that got the highest mean average is:\n")
print(gender_max) 
print(" ")

print("Lastly, for the hometown category, among the three, the hometown that got the highest mean average is:\n")
print(hometown_max) 
```
